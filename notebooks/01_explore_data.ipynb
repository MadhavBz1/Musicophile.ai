```python
audio_features = [
    "Tempo",
    "loudness",
    "Energy",
    "Danceability",
    "Positiveness",
    "Speechiness",
    "Liveness",
    "Acousticness",
    "Instrumentalness"
]
lyrics_column = "text"
meta_features = [
    "Genre",
    "emotion",
    "Key",
    "Explicit"
]
df.head(3)
```

```python
import re
import numpy as np

df = df.rename(columns={
    "Artist(s)": "artist",
    "song": "song",
    "text": "lyrics",
    "Length": "length",
    "Release Date": "release_date",
    "Loudness (db)": "loudness"
})



def parse_loudness(x):
    if pd.isna(x):
        return np.nan
    s = str(x).lower().strip()
    s = s.replace("−", "-").replace("–", "-").replace("—", "-")
    s = s.replace(",", ".")
    m = re.search(r"-?\d+(\.\d+)?", s)
    return float(m.group(0)) if m else np.nan

df["loudness"] = df["loudness"].apply(parse_loudness)

```

```python

numeric_cols = [
    "Tempo", "loudness", "Energy", "Danceability",
    "Positiveness", "Speechiness", "Liveness",
    "Acousticness", "Instrumentalness", "Popularity"
]

other_numeric = [c for c in numeric_cols if c != "loudness"]
for col in other_numeric:
    df[col] = pd.to_numeric(df[col], errors="coerce")

    

def length_to_seconds(x):
    try:
        m, s = x.split(":")
        return int(m) * 60 + int(s)
    except:
        return None


df["length_sec"] = df["length"].apply(length_to_seconds)
df.head(3)

```

```python
import re
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler


def clean_numeric(x):
    if pd.isna(x):
        return np.nan
    x = str(x).lower().strip()
    # remove units like "db", "%" and any other non-numeric chars
    x = re.sub(r"[^0-9.\-]", "", x)
    return float(x) if x not in ("", "-", ".", "-.") else np.nan

for col in audio_features:
    df[col] = df[col].apply(clean_numeric)

df = df.dropna(subset=["lyrics_clean", "Tempo", "Energy", "Danceability"]).reset_index(drop=True)
print("After safe drop:", df.shape)

for col in audio_features:
    df[col] = df[col].fillna(df[col].median())

scaler = StandardScaler()
X_audio = scaler.fit_transform(df[audio_features].values)
X_audio.shape

df[audio_features].head(5)
df[audio_features].isna().mean().sort_values(ascending=False)
```

```python
X_meta = pd.get_dummies(df[meta_features].fillna("UNK"), drop_first=False).values

X_meta.shape
```

```python
import numpy as np
from sklearn.metrics.pairwise import cosine_similarity

def recommend(
    seed_idx,
    w_audio=0.4,
    w_lyrics=0.4,
    w_meta=0.2,
    top_k=10,
    diversify_genre=0.0
):
    # normalize weights
    s = w_audio + w_lyrics + w_meta
    w_audio, w_lyrics, w_meta = w_audio/s, w_lyrics/s, w_meta/s

    sim_audio = cosine_similarity(X_audio[seed_idx:seed_idx+1], X_audio).ravel()
    sim_lyrics = cosine_similarity(X_lyrics[seed_idx], X_lyrics).ravel()
    sim_meta  = cosine_similarity(X_meta[seed_idx:seed_idx+1], X_meta).ravel()

    score = w_audio*sim_audio + w_lyrics*sim_lyrics + w_meta*sim_meta

    if diversify_genre > 0:
        seed_genre = df.loc[seed_idx, "Genre"]
        score -= diversify_genre * (df["Genre"] == seed_genre).astype(float)

    score[seed_idx] = -np.inf
    top_idx = np.argsort(score)[::-1][:top_k]

    return df.loc[top_idx, ["song", "artist", "Genre", "emotion"]].assign(
        score=score[top_idx],
        audio_sim=sim_audio[top_idx],
        lyric_sim=sim_lyrics[top_idx],
        meta_sim=sim_meta[top_idx]
    )
```
